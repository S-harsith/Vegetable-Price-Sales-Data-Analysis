```python
from pyspark.sql.functions import (
    col,
    trim,
    when,
    concat_ws,
    lit
)

cleaned_df = (
    df
    .withColumn("market", trim(col("market")))
    .withColumn("crop_name", trim(col("crop_name")))
    .withColumn(
        "invalid_quantity",
        col("quantity") < 0
    )
    .withColumn(
        "invalid_price",
        (col("min_price") < 0) |
        (col("max_price") < 0) |
        (col("modal_price") < 0) |
        (col("min_price") > col("max_price")) |
        (col("modal_price") < col("min_price")) |
        (col("modal_price") > col("max_price"))
    )
    .withColumn(
        "is_valid",
        (~col("invalid_quantity")) & (~col("invalid_price"))
    )
    .withColumn(
        "quality_issue",
        concat_ws(
            "; ",
            when(col("invalid_quantity"), lit("Negative quantity")),
            when(col("invalid_price"), lit("Invalid price relationship"))
        )
    )
)

display(cleaned_df)
```
